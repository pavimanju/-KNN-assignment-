```python
from sklearn.model_selection import train_test_split

# Features
X = df.drop(columns=["animal_name", "class_type"])

# Target
y = df["class_type"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Features
X = df.drop(columns=["animal_name", "class_type"])

# Target
y = df["class_type"]

# 80% Training, 20% Testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
print("Training data shape:", X_train_scaled.shape)
print("Testing data shape:", X_test_scaled.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn.fit(X_train_scaled, y_train)

print("KNN model trained successfully!")
```

KNN Implementation:
The K-Nearest Neighbours classification algorithm was implemented using Scikit-learn. Feature scaling was performed using StandardScaler because KNN is a distance-based algorithm. A K value of 5 was selected initially, and Euclidean distance was used as the distance metric.

Choice of K and Distance Metric:

K = 5
Distance metric = Euclidean distance
Scaling method = StandardScaler

```python
y_pred = knn.predict(X_test_scaled)

print("Prediction completed!")
print("Predicted values:")
print(y_pred)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-Score :", f1)
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

Model Evaluation:
The KNN classifier was evaluated on the testing dataset using accuracy, precision, recall, and F1-score. These metrics were used to measure the classification performance of the model. Weighted averaging was used because the dataset contains multiple animal classes.

Metrics:

Accuracy: Overall percentage of correctly classified animals.
Precision: Measures how many predicted positive classifications were actually correct.
Recall: Measures how many actual instances were correctly identified.
F1-score: Harmonic mean of precision and recall.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

# Select two features for visualization
X_2d = df[["legs", "catsize"]]
y_2d = df["class_type"]

# Train KNN using two features
knn_2d = KNeighborsClassifier(
    n_neighbors=5,
    metric="euclidean"
)

knn_2d.fit(X_2d, y_2d)

# Create mesh grid
x_min = X_2d["legs"].min() - 1
x_max = X_2d["legs"].max() + 1

y_min = X_2d["catsize"].min() - 1
y_max = X_2d["catsize"].max() + 1

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.02),
    np.arange(y_min, y_max, 0.02)
)

# Predict each point in the grid
Z = knn_2d.predict(
    np.c_[xx.ravel(), yy.ravel()]
)

Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.figure(figsize=(10, 6))

plt.contourf(xx, yy, Z, alpha=0.3)

plt.scatter(
    X_2d["legs"],
    X_2d["catsize"],
    c=y_2d,
    edgecolors="black"
)

plt.xlabel("Number of Legs")
plt.ylabel("Cat Size")
plt.title("KNN Decision Boundary")

plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import matplotlib.pyplot as plt

# Different K values
K_values = [1, 3, 5, 7, 9]

k_results = []

for k in K_values:
    model = KNeighborsClassifier(
        n_neighbors=k,
        metric="euclidean"
    )

    model.fit(X_train_scaled, y_train)

    prediction = model.predict(X_test_scaled)
    accuracy = accuracy_score(y_test, prediction)

    k_results.append([k, accuracy])

# Results table
k_results_df = pd.DataFrame(
    k_results,
    columns=["K", "Accuracy"]
)

print("K Value Comparison Results")
display(k_results_df)

# Best K
best_row = k_results_df.loc[
    k_results_df["Accuracy"].idxmax()
]

print("Best K:", best_row["K"])
print("Best Accuracy:", best_row["Accuracy"])

# K comparison graph
plt.figure(figsize=(7, 5))
plt.plot(
    k_results_df["K"],
    k_results_df["Accuracy"],
    marker="o"
)

plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("KNN K Value Comparison")
plt.ylim(0, 1.05)
plt.grid(True)
plt.show()
```

K Value Selection

Different K values (1, 3, 5, 7, and 9) were tested using the Euclidean distance metric. The model achieved 100% accuracy for K = 1, 3, 5, and 7, while K = 9 achieved 80% accuracy. Since K = 1 achieved the highest accuracy, it was selected as the best K value for the KNN model.

Interview Questions
1. What are the key hyperparameters in KNN?

Answer:
The main hyperparameters in KNN are:

n_neighbors (K) – Number of nearest neighbours considered for prediction.
metric – Distance measure used to find nearest neighbours.
weights – Determines how much importance is given to each neighbour.
2. What distance metrics can be used in KNN?

Answer:
Common distance metrics used in KNN are:

Euclidean distance – Straight-line distance between points.
Manhattan distance – Sum of absolute differences between coordinates.
Minkowski distance – Generalized form of Euclidean and Manhattan distance.
Chebyshev distance – Maximum absolute difference between coordinates.

For this assignment:
We used Euclidean distance with K = 5.
